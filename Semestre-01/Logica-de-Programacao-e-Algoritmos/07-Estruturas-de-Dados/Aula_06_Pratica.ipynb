# 6. Estruturas de Dados - Listas, Tuplas, Dicionários e Strings

Até agora armazenávamos valores únicos em variáveis simples. Quando precisamos gerenciar coleções de dados (como uma lista de produtos, notas de uma turma ou cadastros), utilizamos as estruturas de dados compostas do Python.

```python
# Listas são coleções ordenadas e mutáveis (podemos alterar, adicionar e remover itens)

linguagens = ['Python', 'C++', 'JavaScript', 'Python']
print(f'Lista original: {linguagens}')

# Adicionando um novo elemento ao final da lista
linguagens.append('Rust')
print(f'Após o append: {linguagens}')

# Acessando elementos por índice (começa em 0)
print(f'Primeira linguagem: {linguagens[0]}')
print(f'Última linguagem: {linguagens[-1]}')
```

```python
# Tuplas são semelhantes às listas, mas são IMUTÁVEIS (não podem ser alteradas após a criação)
# Utilizadas para garantir a integridade de dados que não devem mudar.

coordenadas_gps = (-26.9189, -49.2201)
print(f'Latitude e Longitude fixa: {coordenadas_gps}')
print(f'Latitude: {coordenadas_gps[0]}')
```

```python
# Dicionários armazenam dados baseados em pares de "chave: valor" (key-value)
# Excelente para modelar registros estruturados, como perfis ou fichas técnicas.

aluno_uninter = {
    'nome': 'Gabriel Henrique Pereira',
    'curso': 'Engenharia de Computação',
    'semestre': 1,
    'matriculado': True
}

print(f'Ficha do Aluno: {aluno_uninter}')
print(f'Nome do Aluno: {aluno_uninter["nome"]}')
print(f'Curso: {aluno_uninter["curso"]}')
```

```python
# Strings podem ser tratadas como sequências de caracteres

curso_nome = 'Engenharia de Computação'

print(f'Tamanho da string: {len(curso_nome)}')
print(f'Tudo maiúsculo: {curso_nome.upper()}')
print(f'Tudo minúsculo: {curso_nome.lower()}')
print(f'Fatiamento (Primeiras 11 letras): {curso_nome[:11]}')
```
