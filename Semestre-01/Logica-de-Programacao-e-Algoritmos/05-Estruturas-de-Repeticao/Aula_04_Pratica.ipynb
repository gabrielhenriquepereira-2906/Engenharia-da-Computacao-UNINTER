
# 4. Algoritmos Iterativos e Estruturas de Repetição

Até o momento, nossos blocos de código executavam apenas uma vez. As estruturas de repetição (ou laços/loops) permitem que um bloco de código seja executado múltiplas vezes de forma automática, economizando linhas e permitindo processar grandes volumes de dados.

```python
# O Laço for (Para)
# Utilizado quando sabemos exatamente quantas vezes queremos repetir um bloco de instrução.

print('Contagem de 1 a 5:')
for i in range(1, 6):
    print(i)

print('Contagem regressiva:')
for t in range(5, 0, -1):
    print(t)
print('Fogo!')
```

```python
# Percorrendo strings e coleções diretamente

palavra = 'UNINTER'
print(f'Analisando a palavra: {palavra}')

for letra in palavra:
    print(f'Letra encontrada: {letra}')
```

```python
# O Laço while (Enquanto)
# Utilizado quando a repetição depende de uma condição lógica que pode mudar durante a execução,
# ideal para situações onde não sabemos o número exato de repetições antecipadamente.

contador = 1

while contador <= 4:
    print(f'Passo número: {contador}')
    contador += 1

print('Laço while finalizado.')
```

```python
# Exemplo Prático: Média de Notas com Validação de Entrada
# Unindo input, while, acumulador e tomada de decisão.

soma_notas = 0
quantidade = 0
nota = 0

print('--- Sistema de Notas (Digite -1 para encerrar) ---')

while nota != -1:
    nota = float(input('Digite a nota do aluno (-1 para sair): '))

    if nota == -1:
        break
    elif nota < 0 or nota > 10:
        print('Nota inválida! Digite um valor entre 0 e 10.')
        continue

    soma_notas += nota
    quantidade += 1

if quantidade > 0:
    media = soma_notas / quantidade
    print(f'Média final da turma: {media:.2f} (Total de alunos avaliados: {quantidade})')
else:
    print('Nenhuma nota válida foi inserida.')
```
