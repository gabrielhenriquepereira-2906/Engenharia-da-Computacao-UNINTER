# 5. Funções e Modularização

Até agora, escrevíamos todo o código em um único bloco de execução linear. À medida que os sistemas crescem, surge a necessidade de dividir problemas complexos em partes menores, reutilizáveis e fáceis de manter. É aqui que entram as funções.

```python
# Criando nossa primeira função com a palavra-chave 'def'
# Uma função encapsula um bloco de código que só é executado quando é "chamado".

def exibir_cabecalho():
    print('========================================')
    print('   SISTEMA ACADÊMICO - ENGENHARIA UNINTER')
    print('========================================')

# Chamando a função
exibir_cabecalho()
```

```python
# Funções que recebem dados de entrada para processamento dinâmico

def saudar_usuario(nome, curso):
    print(f'Olá, {nome}! Seja bem-vindo(a) aos estudos de {curso}.')

saudar_usuario('Gabriel', 'Engenharia de Computação')
saudar_usuario('Carlos', 'Análise de Sistemas')
```

```python
# Funções que realizam cálculos e devolvem um resultado para o programa principal

def calcular_media(n1, n2, n3):
    soma = n1 + n2 + n3
    media = soma / 3
    return media

# Armazenando o retorno da função em uma variável
resultado_final = calcular_media(8.5, 7.0, 9.2)
print(f'A média calculada pela função é: {resultado_final:.2f}')
```

```python
# Exemplo Prático: Modularizando a tomada de decisão

def verificar_maioridade(idade):
    if idade >= 18:
        return 'Maior de idade (Apto para CNH).'
    else:
        return 'Menor de idade.'

# Testando a função com diferentes valores
idade_usuario = 18
status = verificar_maioridade(idade_usuario)

print(f'Idade informada: {idade_usuario} anos -> Status: {status}')
```
