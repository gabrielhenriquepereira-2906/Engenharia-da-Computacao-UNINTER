
# 3. Estruturas de Seleção ou Decisão
Até agora, nossos algoritmos seguiam uma execução estritamente sequencial (de cima para baixo). Porém, programas reais precisam tomar decisões baseadas em condições. É aqui que entram os desvios condicionais.

```python
# O Comando if (Se)
# O bloco indentado só é executado se a condição for verdadeira (True).

idade = 18

if idade >= 18:
    print('Você é maior de idade. Pode tirar a CNH!')

print('Fim do programa.')
```

```python
# Estrutura Composta: if-else (Se / Senão)
# Se a condição for falsa, o fluxo desvia para o bloco 'else'.

nota = 6.5

if nota >= 7.0:
    print('Aprovado! Parabéns!')
else:
    print('Reprovado. Necessário exame final.')
```

```python
# Estrutura Encadeada: if - elif - else (Múltiplas escolhas)
# Ideal para testar várias faixas de valores ou opções.

clima = 'chuvoso'

if clima == 'ensolarado':
    print('Vou para o parque de bicicleta.')
elif clima == 'chuvoso':
    print('Vou ficar em casa programando e jogando!')
elif clima == 'nublado':
    print('Vou dar uma caminhada leve.')
else:
    print('Clima não reconhecido.')
```

```python
# Exemplo Prático: Cálculo de Índice de Massa Corporal (IMC)
# Unindo operadores aritméticos, entrada de dados e estruturas de seleção.

peso = float(input('Digite seu peso em kg (ex: 70.5): '))
altura = float(input('Digite sua altura em metros (ex: 1.75): '))

imc = peso / (altura ** 2)

print(f'Seu IMC calculado é: {imc:.2f}')

if imc < 18.5:
    print('Classificação: Abaixo do peso normal.')
elif 18.5 <= imc < 25.0:
    print('Classificação: Peso normal (Parabéns!).')
elif 25.0 <= imc < 30.0:
    print('Classificação: Excesso de peso (Sobrepeso).')
else:
    print('Classificação: Obesidade.')
```
